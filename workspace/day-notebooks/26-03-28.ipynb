```python
from adaptive_latents import datasets
from adaptive_latents import CenteringEstimator, KernelSmoother, Pipeline
import numpy as np
import matplotlib.pyplot as plt
```

```python
d = datasets.Zong22Dataset(sub_dataset_identifier=2)
```

```python
p = Pipeline([
    CenteringEstimator(),
    KernelSmoother(tau=.688),
])

centered_and_smoothed_data = p.offline_run_on(d.neural_data)
```

```python
u,s,vh = np.linalg.svd(centered_and_smoothed_data - centered_and_smoothed_data.mean(axis=0), full_matrices=False)
evals = s**2

fig, axs = plt.subplots(ncols=2, figsize=(10,5))
axs[0].plot(evals/evals.sum())
axs[1].plot(np.hstack([[0], np.cumsum(evals)/evals.sum()]))
axs[1].set_ylim(0,1)

axs[0].axvline(6,color='k')
axs[1].axvline(7,color='k')
```
