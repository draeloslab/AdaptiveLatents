```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import pathlib
```

```python
files = sorted(pathlib.Path('.').resolve().glob('*.csv'))
files
```

```python

df = pd.read_csv(files[-1]).drop(columns=['Unnamed: 0']) * 1000
stim_reg_updated = df['stim_reg_updated']
df.drop(columns=['stim_reg_updated'], inplace=True)
df['remainder'] = df['per_loop'] - df.drop(columns=['per_loop']).sum(axis=1)
df['pred_autoreg'] = df.prediction[stim_reg_updated == 0]
df['pred_stimreg']= df.prediction[stim_reg_updated != 0]
df = df[df.columns[np.argsort(df.max(axis=0))[::-1]]]

print(df.per_loop.sum())

df = df.loc[1000:,:]
```

```python
x = np.array(df['per_loop']).copy()
s = x > 150
print(s.sum())
x[s] = np.nan
plt.plot(x)
plt.axhline(1, linestyle='--', color='black')
```

```python
df.describe()
```

```python
results = {}
for column, new_name, threshold in [('dimension_reduction', 'dimension reduction', 0), ('stim_design', 'stimulus optimization', 1), ('pred_autoreg', 'dynamical model update', 0), ('pred_stimreg', 'stimulus effect regression update', 0), ]:
    result = df.loc[df[column] > threshold,column].describe(percentiles=[.25, .5, .75, .98, .99])
    results[new_name] = result

print(pd.DataFrame(results).to_markdown())
```
