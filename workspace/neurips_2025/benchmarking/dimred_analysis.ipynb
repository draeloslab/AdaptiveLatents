```python
import pandas as pd
import json
# import seaborn as sns
import matplotlib.pyplot as plt
from adaptive_latents import datasets
import numpy as np
```

```python
m = datasets.Odoherty21Dataset().neural_data.shape[1]
```

```python
df = []
with open('results.txt') as fh:
    for line in fh:
        df.append(json.loads(line))
df = pd.DataFrame(df)

df['log_time_per_datapoint'] = np.log10(df['total_time'] / m) - 9
```

```python
df.columns
```

```python
%matplotlib qt
fig, ax = plt.subplots()
ax.scatter(df['n'], df['log_time_per_datapoint'], c=df['time_of_save'])
```
