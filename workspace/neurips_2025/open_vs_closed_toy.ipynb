```python
import importlib
import optimization_comparison
import numpy as np
from adaptive_latents import ArrayWithTime
importlib.reload(optimization_comparison)
```

```python
N = 10

rng = np.random.default_rng(4)
```

```python

from adaptive_latents.input_sources.lds_simulation import LDS
all_srs = []
for _ in range(N):
    lds = LDS.circular_lds(rng=rng, obs_d=130)
    _, data, _ = lds.simulate(int((30+1/np.pi) * 80), rng=rng, initial_state=np.array([20, 0]))
    t = np.arange(data.shape[0]) * 1/lds.transitions_per_rotation
    data = ArrayWithTime(data,t)

    srs = optimization_comparison.make_srs(data=data, rng=rng, comparison_preset='optim_open_vs_closed_toy', n_runs=1, show_tqdm=True, overrides=dict(last_dim_red='prosvd'))
    all_srs.append(srs)
srs = {k: [sub_srs[k][0] for sub_srs in all_srs] for k in srs.keys()}
```

```python


proportions, preq_errors, v_delta_errors, s_delta_errors = optimization_comparison.extract_metrics(srs, preq_cutoff=None)
fig = optimization_comparison.open_v_closed_plot(srs, proportions, preq_errors, v_delta_errors, s_delta_errors, show_individuals=False)

```

```python
fig.savefig(f'open_vs_closed_toy.svg', bbox_inches="tight")
```
