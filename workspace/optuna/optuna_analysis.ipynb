```python
import optuna
import pandas as pd
import seaborn as sns
```

```python
study_name = "example"  # Unique identifier of the study.
storage_name = f"mysql://jgould:password@127.0.0.1/{study_name}" # DROP DATABASE example
study = optuna.load_study(study_name=study_name, storage=storage_name)

df = study.trials_dataframe()

df['1-step err'] = df['values_0']
df['log_pred_p'] = df['values_1']
df['entropy'] = df['values_2']

```

```python
%matplotlib qt
g = sns.PairGrid(df, vars=['1-step err', 'log_pred_p', 'entropy'])
g.map_diag(sns.histplot)
g.map_offdiag(sns.scatterplot)
```
